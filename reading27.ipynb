```python

import pandas as pd
```

```python
data = {
    "Name": ["Asha", "Raj", "Meena", "Vikram", "Priya"],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Salary": [35000, 42000, 38000, 50000, 45000],
    "Age": [23, 28, 25, 32, 27]
}

df = pd.DataFrame(data)
df.to_excel("employee_data.xlsx", index=False)

print("Excel file created successfully!")
```

```python
df = pd.read_excel("employee_data.xlsx")
df
```

```python
print("Number of rows and columns:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nBasic statistics:")
print(df.describe())
```

```python
# Check for missing values
print("Missing values:")
print(df.isnull().sum())

# Remove extra spaces from text columns
df["Name"] = df["Name"].str.strip()
df["Department"] = df["Department"].str.strip()

# Make department names consistent
df["Department"] = df["Department"].str.title()

print("\nCleaned data:")
print(df)
```

```python
# Give each employee a 10% salary increase
df["Updated_Salary"] = df["Salary"] * 1.10

# Add a column to classify salaries
df["Salary_Category"] = df["Salary"].apply(
    lambda salary: "High" if salary >= 45000 else "Regular"
)

print(df)
```

```python
# Export the processed data to Excel
df.to_excel("processed_employee_data.xlsx", index=False)

print("Processed Excel output file saved successfully!")
```
